```python
from datetime import datetime
import random

print("=" * 70)
print("🤖 Welcome to Areeba's AI Chatbot")
print("DecodeLabs - Project 1: Rule-Based AI Chatbot")
print("Built using Python Control Flow and Decision Logic")
print("Type 'help' to see commands | Type 'exit' to close")
print("=" * 70)

name = input("\nEnter your name: ").title()

print(f"\nHello {name}! 😊")
print("I am ready to answer your questions.\n")


while True:

    user = input(f"{name}: ").strip().lower()



    if user in ["exit", "bye", "quit", "goodbye"]:
        print("🤖 Bot: Goodbye! Keep learning AI. 🚀")
        break


    elif any(word in user for word in ["hi", "hello", "hey", "assalamualaikum"]):
        print(random.choice([
            f"🤖 Bot: Hello {name}! How can I help you?",
            f"🤖 Bot: Hi {name}! Nice to meet you 😊",
            f"🤖 Bot: Welcome {name}! Let's learn AI together."
        ]))


    # About Bot
    elif "your name" in user or "who are you" in user:
        print("🤖 Bot: I am Areeba's AI Chatbot, a Rule-Based AI project created using Python.")


    elif "created you" in user or "developer" in user:
        print("🤖 Bot: I was created by Areeba Kashif during the DecodeLabs Artificial Intelligence Internship.")


  
    elif "artificial intelligence" in user or user == "ai" or "what is ai" in user:
        print("🤖 Bot: Artificial Intelligence enables machines to perform tasks that normally require human intelligence.")


    elif "machine learning" in user:
        print("🤖 Bot: Machine Learning is a branch of AI where computers learn patterns from data.")


    elif "deep learning" in user:
        print("🤖 Bot: Deep Learning uses neural networks to solve complex problems like image and speech recognition.")


    elif "generative ai" in user:
        print("🤖 Bot: Generative AI creates new content like text, images, audio, and code.")


    elif "llm" in user or "large language model" in user:
        print("🤖 Bot: LLMs are AI models trained on large amounts of text to understand and generate human-like language.")


    elif "nlp" in user or "natural language processing" in user:
        print("🤖 Bot: NLP helps computers understand and process human language.")


    elif "computer vision" in user:
        print("🤖 Bot: Computer Vision allows machines to understand images and videos.")


    
    elif "python" in user:
        print("🤖 Bot: Python is a popular programming language used in AI, Data Science, Automation and Web Development.")


    elif "java" in user:
        print("🤖 Bot: Java is an object-oriented programming language used for software development.")


    elif "c++" in user:
        print("🤖 Bot: C++ is used for system programming, games, and performance-based applications.")


    elif "javascript" in user:
        print("🤖 Bot: JavaScript is mainly used for interactive web development.")


    elif "html" in user:
        print("🤖 Bot: HTML is used to create the structure of web pages.")


    elif "css" in user:
        print("🤖 Bot: CSS is used for designing and styling websites.")


  
    elif "oop" in user:
        print("🤖 Bot: OOP stands for Object-Oriented Programming.")


    elif "inheritance" in user:
        print("🤖 Bot: Inheritance allows one class to acquire properties and methods of another class.")


    elif "polymorphism" in user:
        print("🤖 Bot: Polymorphism allows objects to take different forms.")


    elif "encapsulation" in user:
        print("🤖 Bot: Encapsulation combines data and methods into a single unit and protects data.")


    elif "abstraction" in user:
        print("🤖 Bot: Abstraction hides unnecessary details and shows only important information.")


    
    elif "array" in user:
        print("🤖 Bot: Array stores multiple values of the same type in a structured way.")


    elif "stack" in user:
        print("🤖 Bot: Stack follows LIFO principle: Last In First Out.")


    elif "queue" in user:
        print("🤖 Bot: Queue follows FIFO principle: First In First Out.")


    elif "linked list" in user:
        print("🤖 Bot: Linked List stores data using connected nodes.")


    elif "algorithm" in user:
        print("🤖 Bot: Algorithm is a step-by-step procedure to solve a problem.")


   
    elif "time" in user:
        print("🤖 Bot: Current time is", datetime.now().strftime("%I:%M:%S %p"))


    elif "date" in user:
        print("🤖 Bot: Today's date is", datetime.now().strftime("%d-%m-%Y"))


   
    elif "calculator" in user or "calculate" in user:

        print("🤖 Bot: Welcome to Calculator")

        num1 = float(input("Enter first number: "))
        operator = input("Enter operator (+,-,*,/): ")
        num2 = float(input("Enter second number: "))


        if operator == "+":
            print("🤖 Bot: Answer =", num1 + num2)

        elif operator == "-":
            print("🤖 Bot: Answer =", num1 - num2)

        elif operator == "*":
            print("🤖 Bot: Answer =", num1 * num2)

        elif operator == "/":
            if num2 != 0:
                print("🤖 Bot: Answer =", num1 / num2)
            else:
                print("🤖 Bot: Cannot divide by zero.")

        else:
            print("🤖 Bot: Invalid operator.")


   
    elif "motivate" in user:
        print(random.choice([
            "🤖 Bot: Believe in yourself. Every expert was once a beginner. 🌟",
            "🤖 Bot: Keep learning and improving every day.",
            "🤖 Bot: Success comes from consistency."
        ]))


    elif "joke" in user:
        print(random.choice([
            "🤖 Bot: Why do programmers prefer dark mode? Because light attracts bugs! 😂",
            "🤖 Bot: Debugging is like finding a needle in a haystack."
        ]))


  
    elif "sad" in user:

        choice = input("🤖 Bot: Want motivation? (yes/no): ").lower()

        if choice == "yes":
            print("🤖 Bot: Every challenge makes you stronger. Keep going! 💙")
        else:
            print("🤖 Bot: I am always here whenever you need help.")


    elif "happy" in user:
        print("🤖 Bot: That's great! Keep smiling 😊")


  
    elif "thank" in user:
        print(f"🤖 Bot: You're welcome {name}! 😊")


  
    elif "help" in user:
        print("""
================ COMMANDS ================

AI:
- what is ai
- machine learning
- deep learning
- generative ai
- nlp
- computer vision

Programming:
- python
- java
- c++
- javascript
- html
- css

CS:
- oop
- inheritance
- polymorphism
- array
- stack
- queue
- linked list

Tools:
- calculator
- date
- time

Fun:
- joke
- motivate

Other:
- who are you
- developer
- exit

==========================================
""")


   
    else:
        print("🤖 Bot: Sorry, I don't understand that yet.")
        print("🤖 Bot: Type 'help' to see available topics.")
```
